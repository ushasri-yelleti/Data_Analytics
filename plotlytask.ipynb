#Laptop Price and Specification Analysis Using Python and Plotly

```python
import pandas as pd
df=pd.read_csv('/content/laptops .csv')
print(df)
```

```python
import plotly.express as px
#line chart
fig_line = px.line(
    df.head(20),
    x=df.head(20).index,
    y='price(in Rs.)',
    markers=True,
    title='Laptop Price Variation',
    labels={'index':'Laptop Number','Price_euros':'Laptop Price'}
)
fig_line.show()
```

```python
#bar chart
fig_bar = px.bar(
    df.head(20),
    x='ram',
    y='price(in Rs.)',
    color='name',
    title='Total Laptop Prices by Company',
    labels={'name':'Laptop Brand','price(in Rs.)':'Total Price'}
)
fig_bar.show()
```

```python
#scatter plot
fig_scatter = px.scatter(
    df,
    x='display(in inch)',
    y='price(in Rs.)',
    color='rating',
    title='Screen Size vs Laptop Price',
    labels={'Inches':'Screen Size','Price_euros':'Laptop Price'}
)
fig_scatter.show()
```

```python
#pie chart
fig_pie = px.pie(
    df,
    names='ram',
    title='Ram Distribution',
    labels={'ram':'ram capacity'}
)
fig_pie.show()
```

```python
#donut chart
fig_donut = px.pie(
    df,
    names='os',
    hole=0.5,
    title='Operating System Distribution',
    labels={'os':'Operating System'}
)
fig_donut.show()
```

```python
# Box Plot
fig_box = px.box(
    df,
    x='ram',
    y='price(in Rs.)',
    color='ram',
    title='Laptop Price Distribution by RAM',
    labels={'ram':'RAM Capacity',
            'price(in Rs.)':'Laptop Price'}
)
fig_box.show()
```

```python
# Violin Plot
fig_violin = px.violin(
    df,
    x='display(in inch)',
    y='price(in Rs.)',
    color='display(in inch)',
    box=True,
    points='all',
    title='Laptop Price Distribution by Display Size',
    labels={'display(in inch)':'Display Size',
            'price(in Rs.)':'Laptop Price'}
)
fig_violin.show()
```
