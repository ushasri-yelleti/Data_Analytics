```python

import pandas as pd
df = pd.read_csv("/content/tmdb_5000_movies.csv")
print(df)
```

```python
print("\nStatistical Summary:")
print(df.describe())
```

```python
#3
print("Columns containing missing values:")
cols_with_missing = df.columns[df.isnull().any()]
print(cols_with_missing.tolist())
print("\nNo. of missing values in each column:")
print(df.isnull().sum())
```

```python
#4
df.drop_duplicates(inplace=True)
print("\nUpdated no. of records:", df.shape)
```

```python
#5
import numpy as np
df['runtime'] = df['runtime'].fillna(df['runtime'].median())
df['budget'] = df['budget'].replace(0, np.nan)
df['revenue'] = df['revenue'].replace(0, np.nan)
df['budget'] = df['budget'].fillna(df['budget'].median())
df['revenue'] = df['revenue'].fillna(df['revenue'].median())
df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')
df['release_date'] = df['release_date'].fillna(df['release_date'].mode()[0])
df['homepage'] = df['homepage'].fillna('Unknown')
df['tagline'] = df['tagline'].fillna('Unknown')
df['overview'] = df['overview'].fillna('Unknown')
print("Missing values after handling:")
print(df.isnull().sum())
```

```python
#6
df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')
df['release_year'] = df['release_date'].dt.year
df['release_month'] = df['release_date'].dt.month
df['release_day'] = df['release_date'].dt.day
print(df[['release_date', 'release_year', 'release_month', 'release_day']].head(10))
```

```python
#7
top10_popular = df.sort_values(by='popularity', ascending=False).head(10)
print(top10_popular[['title', 'popularity']])
```

```python
#8
top10_rated = df[df['vote_count'] > 100].sort_values(by='vote_average', ascending=False).head(10)
print(top10_rated[['title', 'vote_average', 'vote_count']])
```

```python
#9
top10_revenue = df.sort_values(by='revenue', ascending=False).head(10)
print(top10_revenue[['title', 'budget', 'revenue']])
```

```python
#10
df['profit'] = df['revenue'] - df['budget']
top10_profitable = df.sort_values(by='profit', ascending=False).head(10)
print(top10_profitable[['title', 'budget', 'revenue', 'profit']])
```

```python
#11
avg_values = df[['budget', 'revenue', 'popularity', 'vote_average']].mean()
print(avg_values)
```
