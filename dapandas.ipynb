```python
#handling duplicates
import pandas as pd
data={
    "empid":[101,102,103,104,105,106,107,108,109,110,110],
    "name":["Ravi","PRIYA","Arun","sneha","Kiran","ANJALI","Vikram","meena","RAHUL","Sneha","Sneha"],
    "city":["Hyderabad","hyderabad","CHENNAI","Chennai","Bangalore","BANGALORE","Mumbai","mumbai","Delhi","Hyderabad","Hyderabad"],
    "salary":[30000,40000,3500,4500,32000,50000,38000,42000,36000,45000,45000]
}
df=pd.DataFrame(data)
print(df)
```

```python
df.duplicated()
```

```python
df[df.duplicated()]
```

```python
df[df.duplicated()].sum()
```

```python
df[df.duplicated(keep=False)]
```

```python
d=df[df.duplicated()]
total=d[["name","salary"]].sum()
print(total)
```
